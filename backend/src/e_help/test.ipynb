```python
import sys
import os
from pathlib import Path
from dotenv import load_dotenv
from langchain.embeddings import init_embeddings

env_path = Path.cwd().parents[1] / ".env"
load_dotenv(env_path, override=True)

print("python:", sys.executable)
print("cwd:", os.getcwd())
print("base_url:", os.getenv("CLOSE_API_BASE_URL"))
print("key_exists:", bool(os.getenv("CLOSE_API_KEY")))
print("env_path:", env_path)
print("backend_env_exists:", env_path.exists())

embeddings = init_embeddings(
    model="text-embedding-3-small",
    provider="openai",
    api_key=os.getenv("CLOSE_API_KEY"),
    base_url=os.getenv("CLOSE_API_BASE_URL"),
)

vector = embeddings.embed_query("你好，很高兴认识你")
print("dimension:", len(vector))
```
